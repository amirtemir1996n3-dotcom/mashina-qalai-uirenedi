## 5.3. Қате функциясы: модель қаншалықты нашар?

```python
import numpy as np

hours  = np.array([1, 2, 2.5, 3, 4, 4.5, 5, 6, 6.5, 7, 8, 9])
scores = np.array([35, 48, 45, 57, 60, 66, 63, 75, 78, 74, 88, 92])

def predict(x, w, b):
    return w * x + b

def mse(y_pred, y_true):
    return np.mean((y_pred - y_true) ** 2)

# Бірнеше түзуді «көзбен» таңдап көрейік
for w, b in [(5, 40), (10, 20), (7, 30)]:
    error = mse(predict(hours, w, b), scores)
    print(f"w = {w:>2}, b = {b}:  MSE = {error:.1f}")
```

## 5.4. Градиенттік түсу

## 5.5. Модельді нөлден оқытамыз

```python
w, b = 0.0, 0.0          # бастапқы мәндер
lr = 0.01                # оқыту қарқыны
n = len(hours)

for epoch in range(3001):
    y_pred = predict(hours, w, b)              # 1. болжам
    error = y_pred - scores                    # 2. қателер
    loss = np.mean(error ** 2)                 #    MSE

    dw = 2 / n * np.sum(error * hours)         # 3. градиент
    db = 2 / n * np.sum(error)

    w = w - lr * dw                            # 4. жаңарту
    b = b - lr * db

    if epoch % 500 == 0:
        print(f"Эпоха {epoch:4d}: қате = {loss:8.2f}, "
              f"w = {w:.3f}, b = {b:.3f}")
```

```python
for h in [0, 5.5, 10]:
    print(f"{h:>4} сағат → болжам: {predict(h, w, b):.0f} ұпай")
```

## 5.7. scikit-learn көмегімен

```python
from sklearn.linear_model import LinearRegression

X = hours.reshape(-1, 1)    # белгілер 2D кесте түрінде болуы керек
model = LinearRegression()
model.fit(X, scores)

print("w =", model.coef_[0].round(3))
print("b =", model.intercept_.round(3))
print("5,5 сағат →", model.predict([[5.5]]).round(1))
```
