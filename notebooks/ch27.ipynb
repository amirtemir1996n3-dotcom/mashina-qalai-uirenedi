```python
def build_prompt(task, role="", context="", fmt="", limits="",
                 examples=()):
    parts = []
    if role:
        parts.append(f"Сен — {role}.")
    parts.append(f"Тапсырма: {task}.")
    if context:
        parts.append(f"Контекст: {context}.")
    if fmt:
        parts.append(f"Жауап форматы: {fmt}.")
    if limits:
        parts.append(f"Шектеулер: {limits}.")
    for i, ex in enumerate(examples, 1):
        parts.append(f"{i}-мысал: {ex}")
    return "\n".join(parts)

physics = build_prompt(
    role="8-сынып оқушыларына физикадан сабақ беретін мұғалім",
    task="Ньютонның үшінші заңын түсіндір",
    context="мен формуланы білемін, бірақ мағынасын түсінбеймін",
    fmt="3 қысқа абзац, соңында 2 тексеру сұрағы",
    limits="күрделі терминдер қолданба, өмірден мысал келтір")
print(physics)
```

```python
import pandas as pd

URL = ("https://raw.githubusercontent.com/amirtemir1996n3-dotcom/"
       "mashina-qalai-uirenedi/main/data/news_kz.csv")
news = pd.read_csv(URL)
shots = news.groupby("санат").sample(1, random_state=4)
fewshot = ("Жаңалық тақырыбын бес санаттың біріне жатқыз:\n"
           "спорт, экономика, технология, мәдениет, денсаулық.\n\n")
for title, cat in shots.values:
    fewshot += f"Тақырып: {title}\nСанат: {cat}\n\n"
fewshot += "Тақырып: Астанада жаңа драма театры ашылды\nСанат:"
print(fewshot)
```

```python
def check_prompt(prompt):
    text = prompt.lower()
    signs = {
        "Рөл": ["сен —", "ретінде"],
        "Контекст": ["мен ", "менің", "үшін"],
        "Формат": ["абзац", "кесте", "тізім", "сөйлем", "формат"],
        "Шектеулер": ["қолданба", "аспасын", "жазба", "тек "],
    }
    for name, words in signs.items():
        found = any(w in text for w in words)
        print(f"{name:10} {'бар' if found else 'жоқ'}")
    n = len(prompt.split())
    print(f"Сөз саны: {n}", "(тым қысқа!)" if n < 8 else "")

check_prompt("Фотосинтез туралы жаз")
print("---")
check_prompt(physics)
```
