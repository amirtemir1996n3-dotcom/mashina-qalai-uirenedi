## 3.4. Тізімдер және сөздіктер

```python
scores = [78, 92, 65, 88, 95]

print(scores[0])       # бірінші элемент
print(scores[-1])      # соңғы элемент
print(scores[1:3])     # 1 және 2 индексті элементтер (кесінді)
print(len(scores))     # элементтер саны

scores.append(70)      # соңына жаңа элемент қосу
print(scores)
print(sum(scores) / len(scores))   # орташа мән
```

## 3.5. Шарттар және циклдер

```python
score = 83

if score >= 85:
    grade = 5
elif score >= 65:
    grade = 4
elif score >= 40:
    grade = 3
else:
    grade = 2

print("Баға:", grade)
```

```python
for s in [78, 92, 65]:
    print(s, "→", "өтті" if s >= 70 else "өтпеді")

total = 0
for i in range(1, 101):     # 1, 2, ..., 100
    total += i              # total = total + i
print("1-ден 100-ге дейінгі сандардың қосындысы:", total)
```

```python
squares = [x ** 2 for x in range(6)]
print(squares)
evens = [x for x in range(20) if x % 2 == 0]
print(evens)
```

## 3.6. Функциялар

```python
def average(numbers):
    """Тізімдегі сандардың орташа мәнін қайтарады."""
    return sum(numbers) / len(numbers)

def bmi(weight, height):
    """Дене салмағының индексі: салмақ (кг) / бой² (м)."""
    return weight / height ** 2

print(average([4, 5, 5, 3, 5]))
print(round(bmi(55, 1.68), 1))
```

## 3.7. NumPy: сандармен жылдам жұмыс

```python
import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print(a + b)        # элементтерді жұптап қосу
print(a * b)        # элементтерді жұптап көбейту
print(a * 3)        # әр элементті 3-ке көбейту
print(a ** 2)       # әр элементті квадраттау
print(a.sum(), a.mean(), a.max())
```

```python
M = np.array([[1, 2, 3],
              [4, 5, 6]])
print(M.shape)          # (жолдар, бағандар)
print(M[0, 2])          # 0-жол, 2-баған
print(M[:, 1])          # 1-бағанның барлық элементі
print(M.T)              # транспонирлеу: жолдар ↔ бағандар
print(M.sum(axis=0))    # әр бағанның қосындысы
```

```python
rng = np.random.default_rng(42)   # 42 — «тұқым» (seed)
print(rng.random(3))              # [0, 1) аралығындағы 3 сан
print(rng.integers(1, 7, size=5)) # 5 рет ойын сүйегін лақтыру
print(rng.normal(0, 1, size=(2, 3)).round(2))
```

## 3.8. Matplotlib: деректерді көзбен көру

```python
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 200)     # 0-ден 10-ға дейін 200 нүкте

plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), "--", label="cos(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Алғашқы графигіміз")
plt.legend()
plt.show()
```

## 3.9. pandas: кестелермен жұмыс

```python
import pandas as pd

df = pd.DataFrame({
    "аты":   ["Әлия", "Арман", "Дана", "Нұрлан"],
    "сағат": [5, 2, 8, 4],
    "ұпай":  [78, 55, 93, 70],
})
print(df)
print("Орташа ұпай:", df["ұпай"].mean())
print(df[df["сағат"] > 3])      # шарт бойынша сүзгілеу
```
