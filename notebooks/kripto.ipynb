## 1-бөлім. Классикалық шифрлар (К1)

### 1.1. Цезарь шифры

```python
AL = "АӘБВГҒДЕЁЖЗИЙКҚЛМНҢОӨПРСТУҰҮФХҺЦЧШЩЪЫІЬЭЮЯ"   # қазақ әліпбиі, 42 әріп
print(len(AL))

def caesar(text, k):
    """Әр әріпті әліпбиде k орын оңға жылжытады; k < 0 болса — солға (шифрды ашу)."""
    out = []
    for ch in text:
        up = ch.upper()
        if up in AL:
            new = AL[(AL.index(up) + k) % 42]
            out.append(new if ch == up else new.lower())
        else:
            out.append(ch)          # бос орын, тыныс белгілері өзгермейді
    return "".join(out)

print(caesar("КІТАП", 3))     # шифрлау
print(caesar("МЮҮВТ", -3))    # ашу
```

```python
qupiya = "ДЯОЯӨ – НЦІ"
for k in range(1, 42):
    print(k, caesar(qupiya, -k))
```

### 1.3. Жиілік талдау: кілтті компьютер өзі табады

Алдымен қазақ мәтініндегі әріптердің жиілігін 1-томның мәтінінен (`data/kitap.txt`) есептейміз. Интернет болмаса,
блокнотқа енгізілген дайын кесте қолданылады.

```python
FREQ_KITAP = {
    "А": 11.56, "Ә": 0.88, "Б": 2.5, "В": 0.11, "Г": 1.1, "Ғ": 1.25, "Д": 5.58,
    "Е": 8.98, "Ё": 0.0, "Ж": 1.79, "З": 1.54, "И": 1.27, "Й": 1.72, "К": 2.73,
    "Қ": 3.09, "Л": 5.11, "М": 3.22, "Н": 6.18, "Ң": 1.51, "О": 2.98, "Ө": 0.95,
    "П": 1.48, "Р": 5.79, "С": 4.2, "Т": 6.0, "У": 1.68, "Ұ": 0.74, "Ү": 0.87,
    "Ф": 0.22, "Х": 0.11, "Һ": 0.0, "Ц": 0.17, "Ч": 0.02, "Ш": 1.26, "Щ": 0.0,
    "Ъ": 0.0, "Ы": 6.96, "І": 5.69, "Ь": 0.33, "Э": 0.07, "Ю": 0.06, "Я": 0.32,
}

from collections import Counter
from urllib.request import urlopen

URL = ("https://raw.githubusercontent.com/amirtemir1996n3-dotcom/"
       "mashina-qalai-uirenedi/main/data/kitap.txt")

def freq(text):
    """Мәтіндегі әр әріптің жиілігі, пайызбен."""
    c = Counter(ch for ch in text.upper() if ch in AL)
    total = sum(c.values())
    return {ch: 100 * c[ch] / total for ch in AL}

try:
    kitap = urlopen(URL, timeout=10).read().decode("utf-8")
    FREQ = freq(kitap)
    print("Жиілік кітаптың мәтінінен есептелді:", len(kitap), "таңба")
except Exception:
    FREQ = FREQ_KITAP
    print("Интернет жоқ: дайын кесте қолданылады")

top = sorted(FREQ.items(), key=lambda x: -x[1])[:8]
print([(ch, round(p, 1)) for ch, p in top])
```

Ұзын шифрмәтінді ұстап алдық. Оның ең жиі әріптерін қазақ тіліндегі ең жиі әріптермен салыстырыңыз.

```python
SHIFR = (
    "Нұлуфсёұгчлв — гңугұгффю дтфип гзгөзгұзгп ңсұжгх фхұгою жюоюө. Икиоёя Ұлөзи Болм Ъиқгұа ғү"
    "ниұл шгффгұюп ғұ ғұяуфя ғояудлзи дяұпиіи сұюпжг кюокюфх гұңюою кгүюұжгп. Өһпзгм ілчұ һқгң "
    "хгңюф үипяөзя дсоюу нтұяпзя, дяұгң фсжюқюпію жгүюұзг гұгд жгоюөю ғо-Нлпзл спю дһқхзюр ксою"
    "п фгуфю. Со ниқ ниоёип фяози нимдяұ ғұяуфиұ кля, го нимдяұя үлұин ниқзиүифяпяп дгмңгзю. Іл"
    "чұөғфяпзиёя ғұяуфиұзя үгпгу, согұзюр кляояёяп фяозяр кляояёяөип үгоюүфюұүг, няоф тқя-гң фг"
    "дюогзю. Дцёяп дяқ зғо сүю фғүяозя нсөуабфиұёи цмұифиөяқ: дгжзгұогөг ғұ няоффя финүиұяу, пғ"
    "флкиүя ңгқгң өғфяпяпи ир нту һңүгмфюп няоффя фгрзгмзю. Дһо өгілпгоюң сңюфхзюр ир ңгұгугмюө"
    " лзивүюөип цпзиү: зиұинфиұзип қгрзюоюң фгхюу, үсо қгрзюоюң дсмюпіг іиіяө ңгдюозгх."
)

F_shifr = freq(SHIFR)
print([(ch, round(p, 1)) for ch, p in sorted(F_shifr.items(), key=lambda x: -x[1])[:8]])
```

Компьютер әр кілтпен шифрды ашып, нәтиженің жиілігін қазақ мәтінінің жиілігімен салыстырады. Айырманы
**хи-квадрат** көрсеткішімен өлшейміз: ол неғұрлым кіші болса, мәтін қазақ мәтініне соғұрлым ұқсас.

```python
def chi2(text):
    c = Counter(ch for ch in text.upper() if ch in AL)
    n = sum(c.values())
    s = 0
    for ch in AL:
        expected = max(FREQ[ch], 0.05) / 100 * n     # сирек әріптерге кішкене «қор» қосамыз
        s += (c[ch] - expected) ** 2 / expected
    return s

def buz(shifr):
    """Әр кілтті тексеріп, ең «қазақшаға ұқсас» нәтиже беретін кілтті қайтарады."""
    baga = {k: chi2(caesar(shifr, -k)) for k in range(42)}
    return min(baga, key=baga.get)

k = buz(SHIFR)
print("Табылған кілт:", k)
print(caesar(SHIFR, -k)[:200], "...")
```

**Тәжірибе.** Шифрмәтіннің тек алғашқы 5, 10, 15, 20, 50 таңбасын алып, `buz()` функциясы қай ұзындықтан бастап дұрыс кілт (4) табатынын тексеріңіз. Қысқа мәтінде неге қателеседі?

```python
for L in [5, 10, 15, 20, 50]:
    print(L, "таңба → кілт:", buz(SHIFR[:L]))
print("«МЕКТЕП» (k = 4) → кілт:", buz(caesar("МЕКТЕП", 4)))
```

### 1.4. XOR шифры

Компьютер мәтінді байттармен (биттермен) сақтайды. XOR амалы: биттер бірдей болса — 0, әртүрлі болса — 1.
Бір кілтпен екі рет XOR-ласақ, мәтін қалпына келеді: (m XOR k) XOR k = m.

**Болжаңыз:** төмендегі кодтың соңғы жолы не шығарады?

```python
m = "ЖИ".encode("utf-8")       # мәтін → байттар
k = 0b10101010                 # кілт: бір байт
c = bytes(b ^ k for b in m)    # ^ — Python-дағы XOR амалы
print("Байттар:   ", m.hex())
print("Шифрмәтін: ", c.hex())
print("Қайта XOR: ", bytes(b ^ k for b in c).decode("utf-8"))
```

## 2-бөлім. Хэш-функциялар (К2)

### 2.1. SHA-256: мәтіннің «саусақ ізі»

```python
import hashlib

def sha(text):
    return hashlib.sha256(text.encode("utf-8")).hexdigest()

a, b = sha("Сәлем"), sha("сәлем")
print(a)
print(b)
print("Ұзындығы:", len(a), "он алтылық таңба =", len(a) * 4, "бит")
```

### 2.2. Көшкін әсері

**Болжаңыз:** бір ғана әріп өзгерсе, хэштің 256 битінің нешеуі өзгереді?

```python
def ozgergen_bit(h1, h2):
    return bin(int(h1, 16) ^ int(h2, 16)).count("1")

print("«Сәлем» мен «сәлем»:", ozgergen_bit(a, b), "бит өзгерді")

sozder = ["алма", "кітап", "мектеп", "дерек", "модель", "нейрон", "шифр", "хэш"]
for s in sozder:
    t = s[:-1] + ("а" if s[-1] != "а" else "е")     # соңғы әріпті ауыстырамыз
    print(f"{s:>8} → {t:<8}", ozgergen_bit(sha(s), sha(t)))
```

### 2.3. Тұзсыз хэш неге қауіпті?

Ойдан шығарылған сайттың дерекқоры «ұрланды» делік: онда құпиясөздердің тұзсыз SHA-256 хэштері сақталған.
Бұзушы жиі қолданылатын құпиясөздердің тізімін алып, олардың хэштерін дерекқормен салыстырады.

```python
derekqor = {                      # ойдан шығарылған пайдаланушылар
    "Айгерім": sha("qwerty"),
    "Нұрлан": sha("123456"),
    "Дана": sha("Kz7!tq#Lm2-Tau"),
}
jii = ["123456", "password", "qwerty", "111111", "123123", "abc123", "000000", "qwerty123", "iloveyou", "admin"]

kesteler = {sha(p): p for p in jii}        # дайын «хэш → құпиясөз» кестесі
for aty, h in derekqor.items():
    print(aty, "→", kesteler.get(h, "табылмады"))
```

### 2.4. Тұзбен хэштеу

Әр пайдаланушыға кездейсоқ **тұз** беріледі. Тұз хэшпен бірге ашық сақталады, бірақ бірдей құпиясөздердің хэшін
әртүрлі етеді. `pbkdf2_hmac` хэшті 100 000 рет қайталап есептейді: бұл бір тексеру үшін байқалмайды, ал миллиондаған
құпиясөзді іріктеуді өте баяулатады.

```python
import os

def tuzdy_hash(qupiyasoz, tuz=None, n=100_000):
    tuz = tuz or os.urandom(16)
    h = hashlib.pbkdf2_hmac("sha256", qupiyasoz.encode("utf-8"), tuz, n)
    return tuz, h

t1, h1 = tuzdy_hash("qwerty")
t2, h2 = tuzdy_hash("qwerty")
print("Тұздар:", t1.hex(), t2.hex())
print("Хэштер бірдей ме?", h1 == h2)

def tekser(qupiyasoz, tuz, h):
    return tuzdy_hash(qupiyasoz, tuz)[1] == h

print("Дұрыс құпиясөз:", tekser("qwerty", t1, h1))
print("Қате құпиясөз: ", tekser("qwerty1", t1, h1))
```

**Қабілетті оқушыларға.** Қайталау санын 10 есе өсіріп (`n=1_000_000`), уақытты өлшеңіз. Баяу хэш неге пайдалы?

```python
import time
for n in [100_000, 1_000_000]:
    t = time.time()
    tuzdy_hash("qwerty", n=n)
    print(n, "қайталау:", round(time.time() - t, 3), "с")
```

## 3-бөлім. Ашық кілт және цифрлық қолтаңба (К3)

### 3.1. RSA шағын сандармен

```python
p, q = 3, 11
n = p * q
phi = (p - 1) * (q - 1)
e = 3
d = pow(e, -1, phi)            # e·d mod phi = 1 болатын d
print("n =", n, " phi =", phi, " d =", d)

m = 4
c = pow(m, e, n)               # шифрлау: m^e mod n
print("Шифрмәтін:", c)
print("Ашылған хабар:", pow(c, d, n))
```

### 3.2. Үлкенірек кілт және сөзді шифрлау

```python
p, q = 1009, 1013
n = p * q
phi = (p - 1) * (q - 1)
e = 65537
d = pow(e, -1, phi)
print("Ашық кілт:", (n, e))

soz = "АЛМА"
shifr = [pow(ord(ch), e, n) for ch in soz]
print("Шифрмәтін:", shifr)
print("Ашылды:", "".join(chr(pow(x, d, n)) for x in shifr))
```

**Байқаңыз:** «А» әрпі екі рет бірдей санға шифрланды. Мұндай шифрды К1-дегідей жиілік талдаумен бұзуға болады. Сондықтан нақты жүйелер хабарды әріптеп шифрламайды және оған кездейсоқ толтырғыш қосады.

### 3.3. Цифрлық қолтаңба

```python
def qoltanba(habar, d, n):
    h = int(sha(habar), 16) % n      # хабардың хэші (n-нен кіші сан)
    return pow(h, d, n)              # жабық кілтпен қол қою

def tekseru(habar, s, e, n):
    h = int(sha(habar), 16) % n
    return pow(s, e, n) == h         # ашық кілтпен тексеру

hat = "Ертең сабақ 9:00-де басталады. Әсем"
s = qoltanba(hat, d, n)
print("Қолтаңба:", s)
print("Түпнұсқа хат:", tekseru(hat, s, e, n))
print("Өзгертілген хат:", tekseru(hat.replace("9:00", "10:00"), s, e, n))
```

### 3.4. Неге нақты RSA-ны бұзу мүмкін емес?

n-ді көбейткіштерге жіктеуге кететін уақытты өлшейміз.

```python
import time

def jikte(n):
    k = 2
    while k * k <= n:
        if n % k == 0:
            return k, n // k
        k += 1
    return n, 1

for n_ in [1009 * 1013, 1000003 * 1000033]:
    t = time.time()
    print(n_, "=", jikte(n_), " уақыты:", round(time.time() - t, 3), "с")
print("Нақты RSA-дағы n — шамамен 617 цифрлы сан: бұл тәсілмен жіктеу ғасырларға созылады.")
```

### 3.5. HTTPS: сайттың сертификаты (интернет керек)

Браузердегі құлып белгісінің артында сайттың сертификаты тұр: оған сенімді орталық цифрлық қолтаңба қойған.

```python
import socket
import ssl

host = "www.gov.kz"
try:
    ctx = ssl.create_default_context()
    with socket.create_connection((host, 443), timeout=5) as sock:
        with ctx.wrap_socket(sock, server_hostname=host) as s:
            cert = s.getpeercert()
    print("Кімге берілген:", dict(x[0] for x in cert["subject"]))
    print("Кім берген:    ", dict(x[0] for x in cert["issuer"]))
    print("Жарамды:       ", cert["notBefore"], "—", cert["notAfter"])
except Exception as e:
    print("Сайтқа қосылу мүмкін болмады:", type(e).__name__)
```
