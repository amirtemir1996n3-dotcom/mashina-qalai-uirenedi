## 10.1. Бірлік — тірлік: нейрондарды біріктіреміз

````python
import numpy as np

def step(z):
    return np.where(z >= 0, 1, 0)

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

# 1-қабат: екі нейрон (OR және AND)
W1 = np.array([[1, 1],
               [1, 1]])
b1 = np.array([-0.5, -1.5])
# 2-қабат: бір нейрон
W2 = np.array([1, -1])
b2 = -0.5

H = step(X @ W1 + b1)       # жасырын қабаттың шығысы
y = step(H @ W2 + b2)       # желінің жауабы
print("Жасырын қабат (OR, AND):\n", H)
print("XOR:", y)
````

## 10.4. Тура таралу: сигнал желі арқылы қалай өтеді

````python
def relu(z):
    return np.maximum(0, z)

def softmax(Z):
    E = np.exp(Z - Z.max(axis=1, keepdims=True))
    return E / E.sum(axis=1, keepdims=True)

rng = np.random.default_rng(0)
W1 = rng.normal(0, 1, (3, 4)); b1 = np.zeros(4)   # 3 кіріс → 4 нейрон
W2 = rng.normal(0, 1, (4, 2)); b2 = np.zeros(2)   # 4 нейрон → 2 шығыс

X = np.array([[0.5, 1.0, -1.0],                   # 1-мысал
              [2.0, 0.0,  1.0]])                  # 2-мысал

Z1 = X @ W1 + b1          # (2, 4)
A1 = relu(Z1)             # (2, 4)
Z2 = A1 @ W2 + b2         # (2, 2)
P = softmax(Z2)           # (2, 2) — ықтималдықтар

print("Пішіндер:", Z1.shape, A1.shape, Z2.shape)
print("Жасырын қабат:\n", A1.round(2))
print("Ықтималдықтар:\n", P.round(3))
````

## 10.5. Белсендіру функциясы неге міндетті?

````python
Wa = rng.normal(size=(3, 4))
Wb = rng.normal(size=(4, 2))
# екі сызықтық қабат бір қабатпен бірдей:
print(np.allclose((X @ Wa) @ Wb, X @ (Wa @ Wb)))
````
