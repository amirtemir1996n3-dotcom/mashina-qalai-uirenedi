```python
import random

# кім кімді жеңеді: тас → қайшыны, қайшы → қағазды, қағаз → тасты
BEATS = {"тас": "қайшы", "қайшы": "қағаз", "қағаз": "тас"}

def winner(player, computer):
    if player == computer:
        return "Тең ойын!"
    if BEATS[player] == computer:
        return "Сіз жеңдіңіз!"
    return "Компьютер жеңді!"

random.seed(2)
for player in ["тас", "қайшы", "қағаз"]:
    computer = random.choice(list(BEATS))
    result = winner(player, computer)
    print(f"Сіз: {player:6s} Компьютер: {computer:6s} → {result}")
```

**Модель файлдары керек.** Teachable Machine-нен экспортталған `keras_model.h5` және `labels.txt` файлдарын, сондай-ақ тексеретін суретті Colab-қа жүктеңіз (сол жақтағы 📁 панелі немесе `files.upload()`).

```python
import numpy as np
from PIL import Image, ImageOps
import tf_keras                      # Colab-та: !pip install tf-keras

model = tf_keras.models.load_model("keras_model.h5", compile=False)
labels = [line.strip().split(" ", 1)[1] for line in open("labels.txt")]

def classify(path):
    image = Image.open(path).convert("RGB")
    image = ImageOps.fit(image, (224, 224))             # өлшемін келтіру
    x = np.asarray(image, dtype=np.float32) / 127.5 - 1 # [-1, 1] аралығы
    probs = model.predict(x[np.newaxis], verbose=0)[0]
    best = int(np.argmax(probs))
    return labels[best], float(probs[best])

gesture, confidence = classify("my_hand.jpg")
print(f"Модель: {gesture} ({confidence:.0%})")
if gesture.lower() in BEATS:
    computer = random.choice(list(BEATS))
    print("Компьютер:", computer, "→", winner(gesture.lower(), computer))
```
