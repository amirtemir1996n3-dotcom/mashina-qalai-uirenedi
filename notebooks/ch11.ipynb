## 11.3. Сандық мысал: бір қадам толығымен

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

x, w, b, y = 2.0, 0.5, -0.5, 1.0

# тура таралу
z = w * x + b
a = sigmoid(z)
L = (a - y) ** 2

# кері таралу (тізбек ережесі)
dL_da = 2 * (a - y)
da_dz = a * (1 - a)
dL_dz = dL_da * da_dz
dL_dw = dL_dz * x
dL_db = dL_dz

# жаңарту
lr = 0.5
w_new = w - lr * dL_dw
b_new = b - lr * dL_db
a_new = sigmoid(w_new * x + b_new)

print(f"Градиенттер: dL/dw = {dL_dw:.4f}, dL/db = {dL_db:.4f}")
print(f"w: {w} → {w_new:.4f},  b: {b} → {b_new:.4f}")
print(f"Жауап: {a:.4f} → {a_new:.4f},  қате: {L:.4f} → {(a_new - y)**2:.4f}")
```

## 11.5. XOR-ды өздігінен үйрететін желі

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
Y = np.array([[0], [1], [1], [0]])

rng = np.random.default_rng(1)
W1 = rng.normal(0, 1, (2, 4)); b1 = np.zeros((1, 4))
W2 = rng.normal(0, 1, (4, 1)); b2 = np.zeros((1, 1))
lr = 1.0

for epoch in range(5001):
    # --- тура таралу ---
    A1 = sigmoid(X @ W1 + b1)
    A2 = sigmoid(A1 @ W2 + b2)
    loss = np.mean((A2 - Y) ** 2)

    # --- кері таралу ---
    dA2 = 2 * (A2 - Y) / len(X)
    dZ2 = dA2 * A2 * (1 - A2)
    dW2 = A1.T @ dZ2
    db2 = dZ2.sum(axis=0, keepdims=True)
    dA1 = dZ2 @ W2.T
    dZ1 = dA1 * A1 * (1 - A1)
    dW1 = X.T @ dZ1
    db1 = dZ1.sum(axis=0, keepdims=True)

    # --- жаңарту ---
    W1 -= lr * dW1; b1 -= lr * db1
    W2 -= lr * dW2; b2 -= lr * db2

    if epoch % 1000 == 0:
        print(f"Эпоха {epoch:4d}: қате = {loss:.4f}")

print("Желінің жауаптары:", A2.round(3).ravel())
```
