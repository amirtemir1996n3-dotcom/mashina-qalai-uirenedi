## 9.3. Белсендіру функциялары

```python
import numpy as np

def step(z):
    return np.where(z >= 0, 1, 0)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def relu(z):
    return np.maximum(0, z)

z = np.array([-2.0, -0.5, 0.0, 0.5, 2.0])
print("step:   ", step(z))
print("sigmoid:", sigmoid(z).round(3))
print("tanh:   ", np.tanh(z).round(3))
print("relu:   ", relu(z))
```

## 9.4. Персептрон

```python
class Perceptron:
    def __init__(self, n_inputs, lr=0.1):
        self.w = np.zeros(n_inputs)    # салмақтар
        self.b = 0.0                   # ығысу
        self.lr = lr                   # оқыту қарқыны

    def predict(self, X):
        return step(X @ self.w + self.b)

    def fit(self, X, y, epochs=20):
        for epoch in range(1, epochs + 1):
            mistakes = 0
            for xi, target in zip(X, y):
                error = target - self.predict(xi)    # 0, +1 немесе -1
                self.w += self.lr * error * xi
                self.b += self.lr * error
                mistakes += int(error != 0)
            if mistakes == 0:
                print(f"{epoch}-эпохада қатесіз үйренді!")
                return self
        print(f"{epochs} эпохадан кейін де қателер бар :(")
        return self

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_and = np.array([0, 0, 0, 1])

p = Perceptron(2).fit(X, y_and)
print("Салмақтар:", p.w, " Ығысу:", round(p.b, 2))
print("Болжамдар:", p.predict(X))
```
