## 18.3. 2-қадам. Мәтін → сандар: «сөздер қапшығы»

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)
print("Матрица пішіні:", X.shape)            # (хабарламалар, сөздер)
print("Сөздіктен үзінді:", vectorizer.get_feature_names_out()[200:206])
```

## 18.4. 3-қадам. Аңғал Байес жіктеуіші

```python
import numpy as np
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import cross_val_score

model = make_pipeline(CountVectorizer(), MultinomialNB())
scores = cross_val_score(model, texts, labels, cv=5)
print("Кросс-валидация:", scores, "→ орташа", scores.mean())

model.fit(texts, labels)       # барлық деректерде оқыту
```

## 18.5. 4-қадам. Жаңа хабарламаларды тексеру

```python
new_messages = [
    "Сіз ұтыс ұтып алдыңыз, сыйлықты алу үшін сілтемені басыңыз",
    "Ертең сабаққа кітапты ұмытпа",
    "Тегін курс: бүгін тіркеліңіз",
    "Анашым, кешке футболдан кейін үйге барамын",
]
probs = model.predict_proba(new_messages)[:, 1]     # спам болу ықтималдығы
for text, p in zip(new_messages, probs):
    label = "СПАМ" if p >= 0.5 else "қалыпты"
    print(f"{p:5.0%}  {label:8s} {text}")
```

## 18.6. 5-қадам. Модель нені үйренді?

```python
nb = model.named_steps["multinomialnb"]
words = model.named_steps["countvectorizer"].get_feature_names_out()
# логарифмдік ықтималдықтардың айырмасы: оң — спамға тән, теріс — қалыпты
ratio = nb.feature_log_prob_[1] - nb.feature_log_prob_[0]

top_spam = np.argsort(ratio)[::-1][:8]
top_ham = np.argsort(ratio)[:8]
print("Спамға тән:   ", ", ".join(words[top_spam]))
print("Қалыптыға тән:", ", ".join(words[top_ham]))
```

## 18.7. 6-қадам. Қазақ тілінің ерекшелігі: жалғаулар

```python
model_char = make_pipeline(
    CountVectorizer(analyzer="char_wb", ngram_range=(2, 4)),
    MultinomialNB())
model_char.fit(texts, labels)

probs_char = model_char.predict_proba(new_messages)[:, 1]
for text, p in zip(new_messages, probs_char):
    print(f"{p:5.0%}  {text}")
```
