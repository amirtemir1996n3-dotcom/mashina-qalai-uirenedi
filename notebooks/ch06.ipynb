## 6.2. Сигмоида: кез келген санды ықтималдыққа айналдыру

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

for z in [-6, -2, 0, 2, 6]:
    print(f"σ({z:>2}) = {sigmoid(z):.3f}")
```

## 6.3. Логистикалық регрессия

```python
rng = np.random.default_rng(1)
n = 60
hours = rng.uniform(0, 10, n).round(1)          # дайындық сағаты
tests = rng.uniform(0, 10, n).round(1)          # жаттығу тесттері
noise = rng.normal(0, 0.9, n)
passed = ((0.55 * hours + 0.45 * tests + noise) > 5.2).astype(int)

X = np.column_stack([hours, tests])             # пішіні (60, 2)
print("X пішіні:", X.shape)
print("Өткендер үлесі:", passed.mean().round(2))
print("Алғашқы 5 оқушы:", X[:5].tolist())
print("Олардың нәтижесі:", passed[:5])
```

## 6.4. Кросс-энтропия: жіктеудің қате функциясы

## 6.5. Логистикалық регрессияны нөлден оқыту

```python
w = np.zeros(2)      # [w1, w2]
b = 0.0
lr = 0.1

for epoch in range(5001):
    p = sigmoid(X @ w + b)                        # ықтималдықтар
    loss = -np.mean(passed * np.log(p + 1e-12)
                    + (1 - passed) * np.log(1 - p + 1e-12))
    grad_w = X.T @ (p - passed) / n               # градиенттер
    grad_b = np.mean(p - passed)
    w -= lr * grad_w
    b -= lr * grad_b
    if epoch % 1000 == 0:
        print(f"Эпоха {epoch:4d}: қате = {loss:.4f}")

print("Салмақтар:", w.round(3), " Ығысу:", round(b, 3))
```

```python
pred = (sigmoid(X @ w + b) >= 0.5).astype(int)
print("Дәлдік:", (pred == passed).mean())

new_students = np.array([[3, 4],     # 3 сағат, 4 тест
                         [7, 6]])    # 7 сағат, 6 тест
probs = sigmoid(new_students @ w + b)
for s, p in zip(new_students, probs):
    print(f"{s[0]} сағат, {s[1]} тест → өту ықтималдығы {p:.0%}")
```

## 6.7. Көпкластық жіктеу және softmax

```python
def softmax(z):
    e = np.exp(z - np.max(z))      # тұрақтылық үшін max-ты азайтамыз
    return e / e.sum()

scores = np.array([2.0, 1.0, 0.1])   # «мысық», «ит», «құс» ұпайлары
print(softmax(scores).round(3), softmax(scores).sum())
```

## 6.8. scikit-learn нұсқасы

```python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression()
clf.fit(X, passed)
print("Дәлдік:", clf.score(X, passed))
print("Ықтималдықтар:", clf.predict_proba([[3, 4], [7, 6]])[:, 1].round(2))
```
