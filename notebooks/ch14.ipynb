## 14.3. Конволюция: суретті сүзгіден өткізу

```python
import numpy as np
import keras

(x_train, y_train), _ = keras.datasets.mnist.load_data()
img = x_train[0] / 255.0                       # «5» цифры, 28×28

def conv2d(image, kernel):
    kh, kw = kernel.shape
    h, w = image.shape
    out = np.zeros((h - kh + 1, w - kw + 1))
    for i in range(out.shape[0]):
        for j in range(out.shape[1]):
            window = image[i:i + kh, j:j + kw]   # сүзгі астындағы аймақ
            out[i, j] = np.sum(window * kernel)
    return out

vertical = np.array([[1, 0, -1],
                     [1, 0, -1],
                     [1, 0, -1]])
horizontal = vertical.T                          # 90°-қа бұрылған сүзгі

edges_v = conv2d(img, vertical)
edges_h = conv2d(img, horizontal)
print("Кіріс:", img.shape, "→ нәтиже:", edges_v.shape)
```

## 14.4. Пулинг: маңыздыны сақтап, кішірейту

```python
def max_pool(image, size=2):
    h, w = image.shape[0] // size, image.shape[1] // size
    blocks = image[:h * size, :w * size].reshape(h, size, w, size)
    return blocks.max(axis=(1, 3))

a = np.array([[1, 3, 2, 0],
              [4, 6, 1, 1],
              [0, 2, 5, 7],
              [1, 1, 3, 2]])
print(max_pool(a))
print("Тік шеттер картасы:", edges_v.shape, "→", max_pool(edges_v).shape)
```

## 14.5. CNN архитектурасы

```python
from keras import layers

model = keras.Sequential([
    keras.Input(shape=(28, 28, 1)),               # 1 — сұр түсті арна
    layers.Conv2D(32, kernel_size=3, activation="relu"),
    layers.MaxPooling2D(pool_size=2),
    layers.Conv2D(64, kernel_size=3, activation="relu"),
    layers.MaxPooling2D(pool_size=2),
    layers.Flatten(),
    layers.Dropout(0.5),
    layers.Dense(10, activation="softmax"),
])
model.summary()
```
