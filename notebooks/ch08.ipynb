## 8.4. Шатасу матрицасы

```python
import numpy as np
from sklearn.metrics import (accuracy_score, precision_score,
                             recall_score, f1_score, confusion_matrix)

# 1 — ауру, 0 — сау
y_true = np.array([1] * 50 + [0] * 150)
y_pred = np.array([1] * 42 + [0] * 8 + [1] * 15 + [0] * 135)

print(confusion_matrix(y_true, y_pred, labels=[1, 0]))
print(f"Дәлдік:    {accuracy_score(y_true, y_pred):.3f}")
print(f"Precision: {precision_score(y_true, y_pred):.3f}")
print(f"Recall:    {recall_score(y_true, y_pred):.3f}")
print(f"F1:        {f1_score(y_true, y_pred):.3f}")
```

## 8.5. Асыра оқыту және жеткіліксіз оқыту

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

X, y = load_breast_cancer(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=0)

print("Тереңдік | Оқыту | Тест")
for depth in [1, 2, 3, 5, 8, None]:
    t = DecisionTreeClassifier(max_depth=depth, random_state=0)
    t.fit(X_tr, y_tr)
    train_acc, test_acc = t.score(X_tr, y_tr), t.score(X_te, y_te)
    print(f"{str(depth):>8} | {train_acc:.3f} | {test_acc:.3f}")
```

## 8.7. Кросс-валидация

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import cross_val_score

X, y = load_iris(return_X_y=True)
for depth in [1, 2, 3, 4]:
    model = DecisionTreeClassifier(max_depth=depth, random_state=0)
    scores = cross_val_score(model, X, y, cv=5)
    print(f"тереңдік {depth}: {scores.round(2)} → орташа {scores.mean():.3f}")
```
