## 4.1. Векторлар: сандар тізімі

```python
import numpy as np

x = np.array([2, 1, 1])        # кірістер
w = np.array([0.5, -1, 3])     # салмақтар

print(x * w)                   # элементтерді көбейту
print(np.sum(x * w))           # қосындысы
print(np.dot(x, w))            # дәл сол, бір функциямен
print(x @ w)                   # @ — көбейтудің қысқа жазылуы
```

## 4.2. Матрицалар: векторлар кестесі

```python
A = np.array([[1, 2, 0],
              [3, 1, 2]])          # 2×3
B = np.array([[2, 1],
              [0, 3],
              [4, 1]])             # 3×2

C = A @ B                          # 2×2
print(C)
print(A.shape, B.shape, "→", C.shape)
```

## 4.3. Туынды: функцияның көлбеулігі

```python
def f(x):
    return x ** 2

def derivative(f, x, h=1e-5):
    """Туындыны сандық әдіспен жуықтап есептеу."""
    return (f(x + h) - f(x)) / h

for x in [-2, 0, 1.5]:
    print(f"x = {x:>4}:  f'(x) ≈ {derivative(f, x):.3f}")
```

## 4.4. Статистика және ықтималдық

```python
rng = np.random.default_rng(3)
heights = rng.normal(165, 7, size=600)   # 600 оқушының бойы

print(f"Орташа:     {heights.mean():.1f} см")
print(f"Ст. ауытқу: {heights.std():.1f} см")
print(f"Ең аласа:   {heights.min():.1f} см")
print(f"Ең биік:    {heights.max():.1f} см")
```

```python
coins = rng.integers(0, 2, size=10_000)   # 0 — сан, 1 — елтаңба
print("Елтаңба үлесі:", coins.mean())

dice = rng.integers(1, 7, size=60_000)
counts = np.bincount(dice)[1:]            # әр санның неше рет түскені
print("Ықтималдықтар:", (counts / len(dice)).round(3))
```
