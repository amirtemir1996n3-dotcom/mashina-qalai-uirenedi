## 1.2. Тензор деген не?

```python
import torch

x = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])
print(x)
print("Пішіні:", x.shape)
print("Өлшем саны:", x.ndim)
print("Деректер түрі:", x.dtype)
```

## 1.3. Тензорларды жасаудың жолдары

```python
torch.manual_seed(0)
a = torch.zeros(2, 3)          # нөлдер
b = torch.ones(3)              # бірлер
c = torch.arange(0, 10, 2)     # 0-ден 10-ға дейін, 2 қадаммен
d = torch.linspace(0, 1, 5)    # 0 мен 1 арасындағы 5 сан
e = torch.rand(2, 2)           # [0; 1) аралығынан кездейсоқ
f = torch.randn(3)             # қалыпты үлестірімнен кездейсоқ
for name, t in [("a", a), ("b", b), ("c", c), ("f", f)]:
    print(name, tuple(t.shape), t.dtype)
print(d)
print(e)
```

```python
x = torch.tensor([1, 2, 3])
y = x.float()                  # float32-ге айналдыру
z = x.to(torch.float16)        # жартылай дәлдік
print(y, y.dtype)
print(z, z.dtype)
print(x / 2)                   # бөлгенде нәтиже бөлшек болады
```

## 1.4. NumPy-мен байланыс

```python
import numpy as np

arr = np.array([1.0, 2.0, 3.0])
t = torch.from_numpy(arr)      # жадты бөліседі
arr[0] = 100.0                 # массивті өзгертсек...
print(t)                       # ...тензор да өзгереді
back = t.numpy()
print(type(back).__name__, back.dtype)
```

## 1.5. Пішінмен жұмыс

```python
x = torch.arange(12)
m = x.reshape(3, 4)
print(m)
print(m.reshape(2, -1).shape)   # -1: «қалғанын өзің есепте»
img = torch.rand(28, 28)
batch = img.unsqueeze(0).unsqueeze(0)   # жаңа өлшемдер қосу
print(batch.shape)
print(batch.squeeze().shape)            # 1-ге тең өлшемдерді алу
print(batch.flatten(start_dim=1).shape) # әр суретті векторға
hwc = torch.rand(32, 32, 3)             # биіктік, ені, арна
chw = hwc.permute(2, 0, 1)              # арна, биіктік, ені
print(chw.shape)
```

```python
a = torch.ones(2, 3)
b = torch.zeros(2, 3)
print(torch.cat([a, b], dim=0).shape)   # жолдар бойынша
print(torch.cat([a, b], dim=1).shape)   # бағандар бойынша
print(torch.stack([a, b]).shape)        # жаңа өлшем
```

## 1.6. Индекстеу және кесу

```python
m = torch.arange(1, 13).reshape(3, 4)
print(m)
print(m[0])          # бірінші жол
print(m[:, 1])       # екінші баған
print(m[1:, 2:])     # төменгі оң бұрыш
print(m[m > 9])      # шарт бойынша таңдау
m[m % 2 == 0] = 0    # жұп сандарды нөлге ауыстыру
print(m)
```

## 1.7. Тензорлармен есептеу

```python
scores = torch.tensor([[80., 95., 70.],
                       [60., 75., 90.]])   # 2 оқушы, 3 пән
print(scores + 5)
print(scores.mean())          # барлық бағалардың орташасы
print(scores.mean(dim=0))     # әр пәннің орташасы
print(scores.mean(dim=1))     # әр оқушының орташасы
print(scores.argmax(dim=1))   # әр оқушының ең жақсы пәні
```

```python
x = torch.tensor([[1., 2., 3.],
                  [4., 5., 6.]])        # (2, 3)
row = torch.tensor([10., 20., 30.])     # (3,)
col = torch.tensor([[100.], [200.]])    # (2, 1)
print(x + row)
print(x + col)
z = (x - x.mean(dim=0)) / x.std(dim=0)  # әр бағанды стандарттау
print(z)
```

```python
W = torch.randn(4, 3)        # 3 кіріс, 4 шығыс
x = torch.randn(3)           # бір мысал
print((W @ x).shape)
X = torch.randn(5, 3)        # 5 мысалдан тұратын топтама
print((X @ W.T).shape)       # әр мысал үшін 4 шығыс
```

## 1.8. Графикалық процессор

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Құрылғы:", device)
x = torch.randn(1000, 1000, device=device)
y = x @ x
print(y.device.type)
```

```python
import time

def matmul_time(device, n=2000, repeats=5):
    a = torch.randn(n, n, device=device)
    start = time.perf_counter()
    for _ in range(repeats):
        b = a @ a
    if device == "cuda":
        torch.cuda.synchronize()   # GPU есептеуді аяқтағанша күту
    return (time.perf_counter() - start) / repeats

print(f"CPU: {matmul_time('cpu'):.3f} с")
if torch.cuda.is_available():
    print(f"GPU: {matmul_time('cuda'):.4f} с")
```
