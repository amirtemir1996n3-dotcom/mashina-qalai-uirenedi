## 5.2. Үлестірім деген не?

```python
import torch
from torch.distributions import Normal

torch.manual_seed(0)
heights = Normal(loc=170.0, scale=7.0)    # бойлардың «моделі»
print(heights.sample((5,)))               # 5 «жаңа оқушы»
p = heights.cdf(torch.tensor(180.0)) - heights.cdf(torch.tensor(175.0))
print(f"175–180 см аралығының ықтималдығы: {p.item():.3f}")
```

## 5.3. Ең үлкен шындыққа ұқсастық

```python
torch.manual_seed(1)
young = Normal(145.0, 6.0).sample((200,))   # 6-сынып
older = Normal(172.0, 7.0).sample((200,))   # 11-сынып
data = torch.cat([young, older])            # біз тек осыны көреміз

mu, sigma = data.mean(), data.std()
print(f"орташа={mu:.1f}, стандартты ауытқу={sigma:.1f}")
model1 = Normal(mu, sigma)
print(f"орташа лог-ұқсастық: {model1.log_prob(data).mean():.3f}")
```

```python
from sklearn.mixture import GaussianMixture

x = data.numpy().reshape(-1, 1)
gm = GaussianMixture(n_components=2, random_state=0).fit(x)
print("орталары:", gm.means_.ravel().round(1))
print("ауытқулары:", (gm.covariances_.ravel() ** 0.5).round(1))
print("салмақтары:", gm.weights_.round(2))
print(f"орташа лог-ұқсастық: {gm.score(x):.3f}")
print("жаңа оқушылар:", gm.sample(5)[0].ravel().round(1))
```
